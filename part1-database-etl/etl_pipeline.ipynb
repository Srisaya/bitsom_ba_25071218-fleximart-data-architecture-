```python
import pandas as pd

customers_df = pd.read_csv("data/customers_raw.csv")
products_df = pd.read_csv("data/products_raw.csv")
sales_df = pd.read_csv("data/sales_raw.csv")


```

```python
print(customers_df.shape)
print(products_df.shape)
print(sales_df.shape)
```

```python
customers_df = customers_df.drop_duplicates()
print(customers_df.shape)
```

```python
sales_df = sales_df.drop_duplicates()
print(sales_df.shape)
```

```python
sales_df.duplicated().sum()
```

```python
customers_df['email'] = customers_df['email'].fillna(
    customers_df['first_name'].str.lower() + "." +
    customers_df['last_name'].str.lower() + "@fleximart.com"
)
```

```python
def clean_phone(phone):
    phone = str(phone)
    phone = phone.replace("+91", "").replace("-", "").lstrip("0")
    return "+91-" + phone[-10:]

customers_df['phone'] = customers_df['phone'].apply(clean_phone)
```

```python
customers_df['registration_date'] = pd.to_datetime(
    customers_df['registration_date'],
    errors='coerce',
    dayfirst=True
)
```

```python
customers_df['city'] = customers_df['city'].str.title()
```

```python
customers_df.head()
customers_df.isnull().sum()
```

```python
customers_df[customers_df['registration_date'].isna()]
```

```python
median_date = customers_df['registration_date'].median()
customers_df['registration_date'] = customers_df['registration_date'].fillna(median_date)
```

```python
customers_df.head()
customers_df.isnull().sum()
```

```python
products_df.isnull().sum()
```

```python
products_df.head()
```

```python
products_df['price'] = products_df.groupby('category')['price']\
.transform(lambda x: x.fillna(x.mean()))

products_df['stock_quantity'] = products_df['stock_quantity'].fillna(0)

products_df['category'] = products_df['category'].str.capitalize()

products_df['product_name'] = products_df['product_name'].str.strip()
```

```python
products_df.isnull().sum()
```

```python
sales_df.isnull().sum()
```

```python
sales_df = sales_df.drop_duplicates()

sales_df = sales_df.dropna(subset=['customer_id', 'product_id'])
```

```python
sales_df['transaction_date'] = pd.to_datetime(
    sales_df['transaction_date'],
    errors='coerce',
    dayfirst=True
)

median_date = customers_df['registration_date'].median()
customers_df['registration_date'] = customers_df['registration_date'].fillna(median_date)
```
